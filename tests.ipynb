## Por dónde empezar

**Última cobertura medida (unión local parcial):** 81,79 % de líneas del frontend y 93,25 % de líneas de la aplicación backend. Consulte el [desglose, método y límites](notebooks/automatizacion.ipynb#cobertura-local-parcial). No es una cifra global ni un resultado de CI.

1. Abra la ficha del caso en el catálogo. Cada ID tiene una sola ficha de referencia.
2. Los criterios del Plan Maestro están en las fichas. Consulte el [cierre de revisión local](notebooks/cierre.ipynb) para alcance y pendientes, y la [matriz maestra](docs/test/qa-master-matrix.md) para las referencias históricas **U/G/F/W/D**.
3. Para revisar los resultados locales, abra `reporte_tests.xml` después de ejecutar las suites. JUnit cuenta suites, no resultados individuales por ID. Use la [guía de pruebas](test/README.md) para reproducir; abrir el índice no ejecuta pruebas.
4. Para reproducir localmente, lea los criterios justo antes de la celda de código: **Plan** indica qué se exige, **Qué comprueba esta celda** describe las aserciones y **Límite** indica qué queda fuera. Algunas celdas ejecutan una suite completa, no solo los casos de esa página.

## Estados de las fichas

**V:** verificado en el alcance descrito. **P:** evidencia incompleta, contradictoria o solo estática. **N:** pendiente, sin prueba observada de la regla. **M:** aceptación del usuario sin ejecución independiente acreditada.

**En estudio** es un subestado de **P**, no de **V**. Se aplica a TC-AUTO-02: hay mediciones, pero no acreditan el umbral de forma reproducible.

Corte local del 22 de noviembre de 2026. Los estados registrados no equivalen a cobertura ejecutable actual: TC-FUNC-03 y TC-FUNC-04 conservan V histórico, pero no tienen una prueba vigente del criterio completo. Consulte las [brechas por ID](notebooks/cierre.ipynb#brechas-de-reproduccion). TC-USAB-02 conserva P pese a las correcciones verificadas: faltan los insumos originales de conformidad. NVDA y las cinco sesiones humanas quedan fuera de esta entrega; los checks en vivo siguen pendientes. La matriz conserva por separado las filas del corte histórico.
