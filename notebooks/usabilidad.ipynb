## TC-USAB-01

- **Criterio:** Accesibilidad y teclado en Login/Registro y barra de navegación
- **Método previsto:** Manual + Lighthouse.
- **Estado:** **P — Parcial**. El foco supera 3:1 en el recorrido medido tras la corrección. Orca emitió ambos errores en los dos tamaños, pero no se ejecutó el protocolo con NVDA.
- **Evidencia:** Ejecución local del 22 de noviembre de 2026, detallada abajo. Se conserva la aceptación histórica del usuario (Engram #936), que no certificaba una ejecución independiente.
- **Brecha o límite:** La medición de locución corresponde a Orca, no a NVDA; no mide la latencia hasta el oído. El anuncio de apertura del menú se registró después de pulsar Tab. La revisión automatizada del contraste de texto en tema claro no acredita por sí sola conformidad integral.

**Procedimiento del Plan (§5.3)**

Protocolo previsto; los resultados de la ejecución parcial se registran después de la ficha.

- **Preparación:** Abrir Login con NVDA activo, sin mouse; usar 1280×800 y 375×667 y la cuenta de prueba indicada en la ficha.
- **Recorrido:** 1. Recorrer navegación y formulario con Tab y Shift+Tab. 2. Revisar el foco de cada enlace, campo y botón. 3. Enviar una contraseña incorrecta con Enter. 4. Repetir en registro con un correo usado. 5. A 375 px, abrir el menú con Enter y cerrarlo con Escape. 6. Ejecutar Lighthouse y axe-core.
- **Éxito:** Foco visible ≥3:1, orden lógico sin trampas; errores anunciados con `role="alert"`; menú con `aria-expanded`, cierre con Escape y devolución de foco; contraste de texto ≥4.5:1, Lighthouse ≥95 y cero violaciones críticas en axe-core.
